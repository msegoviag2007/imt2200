### ***Parte I: preparación y funciones base***

Antes de responder las preguntas, prepare los datos de modo que las variables utilizadas sean consistentes y analizables. Como mínimo:

a) Cargue las hojas necesarias del archivo Excel.

```python
# Carga inicial de librerías (pandas, numpy, matplotlib.pyplot) a usar.
import pandas as pd, numpy as np, matplotlib.pyplot as plt

# Función auxiliar para asignar a cada hoja su nombre.
def etiquetar_hoja(nombre_hoja: str, df: pd.DataFrame) -> pd.DataFrame:
    try:
        df.name = nombre_hoja
        print(f'[OK] HOJA: {nombre_hoja} ha sido cargada correctamente :)')
        return df
    except Exception as e:
        print(f'[ERROR] HOJA: {nombre_hoja} no ha podido cargarse :(')
        print(f'Detalle del error: {e}')
    return
        
# Lectura de ruta, apertura de archivo y guardado de hojas.
ruta = './Mundial_2026_Tarea1.xlsx'
excel = pd.read_excel(ruta, sheet_name = None)
print('< CARGA DE HOJAS NECESARIAS >')
hojas_a_cargar = ['Partidos', 'Estadisticas_Equipo_Partido', 'Equipos', 'Estadios', 'Arbitros']
df_partidos, df_eep, df_equipos, df_estadios, df_arbitros = [etiquetar_hoja(nombre_hoja, df) for nombre_hoja, df in excel.items() if nombre_hoja in hojas_a_cargar]
```

b) Revise dimensiones, nombres de columnas, tipos de datos y valores faltantes.

```python
# Lectura de hojas de manera individual para verificar funcionamiento.
print('< RESUMEN DE HOJAS >')
dataframes = [df_partidos, df_eep, df_equipos, df_estadios, df_arbitros]
for i, df in enumerate(dataframes):
    print(f'=== HOJA: {df.name} ===')
    print(f'* Filas: {df.shape[0]}, Columnas: {df.shape[1]}')
    print(f'* Primeras 5 columnas: {list(df.columns[:5])}')
    if i < len(dataframes) - 1:
        print()
```
